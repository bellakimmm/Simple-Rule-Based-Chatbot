```python
print("Welcome to Simple Chatbot!")
print("Type 'help' to see what I can do.")
print("Type 'bye' to exit.")

while True:

    user = input("You: ").lower()

    if user == "hello":
        print("Bot: Hello! How are you today?")

    elif user == "help":
        print("Bot: I can respond to:")
        print("- hello")
        print("- help")
        print("- time")
        print("- name")
        print("- bye")

    elif user == "time":
        import datetime
        print("Bot:", datetime.datetime.now())

    elif user == "name":
        print("Bot: I am a simple chatbot created in Google Colab.")

    elif user == "bye":
        print("Bot: Goodbye!")
        break

    else:
        print("Bot: Sorry, I don't understand that question.")
```
